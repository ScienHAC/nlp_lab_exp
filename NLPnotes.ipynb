## 2. Tokenization

**Tokenization** is the process of breaking a stream of text into words, phrases, symbols, or other meaningful elements called **tokens**. The list of tokens then becomes the input for further processing.

There are two main types of tokenization:

### a) Word Tokenization

Splitting text into individual words. NLTK provides `word_tokenize` for this purpose.

```python
import nltk
from nltk.tokenize import word_tokenize

# Download punkt_tab instead of punkt
nltk.download('punkt_tab')

text = "NLTK is a powerful library for Natural Language Processing."
words = word_tokenize(text)
print(f"Original Text: {text}")
print(f"Word Tokens: {words}")

text_two = "I'm starting my AI journey seriously, and I want core depth."
words_two = word_tokenize(text_two)
print(f"\nOriginal Text 2: {text_two}")
print(f"Word Tokens 2: {words_two}")
```

Notice how `word_tokenize` intelligently handles punctuation and contractions (like "I'm" becoming "I" and "'m").

### b) Sentence Tokenization

Splitting text into individual sentences. NLTK provides `sent_tokenize` for this.

```python
import nltk
from nltk.tokenize import sent_tokenize

# Download the required tokenizer data
nltk.download('punkt_tab')

long_text = "This is the first sentence. It has some punctuation. Is this the second sentence? Yes, it is!"
sentences = sent_tokenize(long_text)

print(f"Original Text: {long_text}")
print(f"Sentence Tokens: {sentences}")
```
